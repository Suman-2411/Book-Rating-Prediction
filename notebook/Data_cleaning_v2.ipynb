The ("  num_pages") column has hidden spaces in the front - it will cause the error when we try to use that
To Overcome this problem we can clean all the column names once - remove spaces and same style for all

```python
df.columns=df.columns.str.strip().str.title()
```

we checked for empty cells and found nothing. but the real missing data is hiding as fake numbers - books with 0 pages and books with 0 ratings, we handle those separately

```python
df.isna().sum()
```

```python
df = df.drop_duplicates(subset=['Bookid']).reset_index(drop=True)
df.shape
```

The Publication date is stored as str(Text) like 9/16/2006, so we take the last 4 characters to get the year. then convert it to a number and calculate the age of the book = 2026 − publication year

```python
df['Publication_Year'] = df['Publication_Date'].str[-4:]
df['Publication_Year'] = pd.to_numeric(df['Publication_Year'])

df['Age']=2026-df['Publication_Year']
```

```python
df['Publication_Year'].min(),df['Publication_Year'].max()
```

the language codes are messy (eng, en-us, en-gb all mean English), so a small function cleans them: 
first it checks the keys — if the word is present,the right value is translated and given back. 
then it checks the values — if the word is present, return it as it is.
or else mark it as Others.

```python
language = {'eng':'English','en-us':'English','en-gb':'English','en-ca':'English',
            'fre':'French','spa':'Spanish','ger':'German','jpn':'Japanese'}

def clean_language(key):
    if key in language:
        return language[key]
    if key in language.values():
        return key
    return 'Others'

df['Language_Code']=df['Language_Code'].str.lower().apply(clean_language)
df['Language_Code'].value_counts()
```

Books with 0 pages and Books with 0 ratings.
A book cannot have 0 pages, so those are data errors → we drop them.
The 0 rating books are real (real books, just nobody rated them) → we keep them here but leave them out later when training the model because their rating is not real.

```python
print("Books with 0 ratings:", (df['Ratings_Count'] == 0).sum())
print("Books with 0 pages:", (df['Num_Pages'] == 0).sum())
```

```python
df =df[(df['Num_Pages'] > 0) & (df['Ratings_Count'] > 0)].reset_index(drop=True)
df.shape
```
